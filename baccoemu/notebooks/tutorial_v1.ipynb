```python
# -*- coding: utf-8 -*-
import numpy as np
import matplotlib.pyplot as plt
import baccoemu

# -------------------------
# Initialize emulator
# -------------------------
emulator = baccoemu.Matter_powerspectrum()

# -------------------------
# Cosmological + nonlinear parameters
# -------------------------
params = {
    'omega_cold'    : 0.315,
    'sigma8_cold'   : 0.83,
    'omega_baryon'  : 0.05,
    'ns'            : 0.96,
    'hubble'        : 0.67,
    'neutrino_mass' : 0.0,
    'w0'            : -1.0,
    'wa'            : 0.0,
    'expfactor'     : 1.0,

    # Nonlinear / baryonic parameters
    'M_c'           : 14,
    'eta'           : -0.3,
    'beta'          : -0.22,
    'M1_z0_cen'     : 10.5,
    'theta_out'     : 0.25,
    'theta_inn'     : -0.86,
    'M_inn'         : 13.4
}

# -------------------------
# Linear power spectrum
# -------------------------
k_lin = np.logspace(-4, np.log10(50), 300)  # valid range for linear emulator
k_lin, pk_lin_cold  = emulator.get_linear_pk(k=k_lin, cold=True,  **params)
k_lin, pk_lin_total = emulator.get_linear_pk(k=k_lin, cold=False, **params)

# -------------------------
# Nonlinear power spectrum
# -------------------------
k_nl_max = emulator.emulator['nonlinear']['k'].max()
k_nl = np.logspace(np.log10(emulator.emulator['nonlinear']['k'].min()),
                   np.log10(k_nl_max), 300)

k_nl, pk_nl_cold  = emulator.get_nonlinear_pk(k=k_nl, cold=True,  baryonic_boost=False, **params)
k_nl, pk_nl_total = emulator.get_nonlinear_pk(k=k_nl, cold=False, baryonic_boost=False, **params)

# -------------------------
# Nonlinear with baryonic corrections (safe k)
# -------------------------
k_safe = k_nl[k_nl <= k_nl_max]
k_safe, pk_nl_total_b = emulator.get_nonlinear_pk(k=k_safe, cold=False, baryonic_boost=True, **params)
_, S = emulator.get_baryonic_boost(k=k_safe, **params)

# -------------------------
# Plot results
# -------------------------
fig, ax = plt.subplots(1,2, figsize=(14,5))

# Left: baryonic boost factor
ax[0].semilogx(k_safe, S, color='green')
ax[0].set_xlabel("k [h/Mpc]")
ax[0].set_ylabel(r"$S(k) = P_{\rm baryons}/P_{\rm gravity\,only}$")
ax[0].set_title("Baryonic Boost Factor")
ax[0].grid(True, which='both', ls='--', alpha=0.5)

# Right: matter power spectra
ax[1].loglog(k_lin, pk_lin_cold,  label="Linear Cold")
ax[1].loglog(k_lin, pk_lin_total, label="Linear Total")
ax[1].loglog(k_nl, pk_nl_cold,    label="Nonlinear Cold")
ax[1].loglog(k_nl, pk_nl_total,   label="Nonlinear Total")
ax[1].loglog(k_safe, pk_nl_total_b, label="Nonlinear Total + Baryons")
ax[1].set_xlabel("k [h/Mpc]")
ax[1].set_ylabel(r"P(k) [h^{-3} Mpc^3]")
ax[1].set_title("Matter Power Spectra")
ax[1].legend()
ax[1].grid(True, which='both', ls='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
