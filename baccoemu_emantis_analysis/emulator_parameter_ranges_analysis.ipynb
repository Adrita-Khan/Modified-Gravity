- # Initialize and display emulator (baccoemu) parameters


```python
import logging
import baccoemu

# Set the logging level to ERROR or CRITICAL to suppress lower-level logs
logging.basicConfig(level=logging.CRITICAL)

# Initialize the matter power spectrum emulator
mpk_emulator = baccoemu.Matter_powerspectrum()

# Initialize the biased tracers emulator
lbias_emulator = baccoemu.Lbias_expansion()

# Function to display parameters in a more readable format
def display_parameters(emulator_type, keys, bounds):
    print(f"\n{emulator_type} Parameters:")
    print("=" * 40)
    print(f"{'Parameter':<30} {'Bounds'}")
    print("-" * 40)
    for key, bound in zip(keys, bounds):
        print(f"{key:<30} {bound}")
    print("=" * 40)

# Display parameters for the linear emulator
display_parameters("Linear Emulator", mpk_emulator.emulator['linear']['keys'], mpk_emulator.emulator['linear']['bounds'])

# Display parameters for the nonlinear emulator
display_parameters("Nonlinear Emulator", mpk_emulator.emulator['nonlinear']['keys'], mpk_emulator.emulator['nonlinear']['bounds'])

# Display parameters for the biased tracers emulator
display_parameters("Biased Tracers Emulator", lbias_emulator.emulator['nonlinear']['keys'], lbias_emulator.emulator['nonlinear']['bounds'])
```

- # Display parameter ranges for NonLinearMGBoostEmulator (e-MANTIS)

```python
from emantis.matter_power_spectrum import NonLinearMGBoostEmulator

# Initialize the emulator
emu = NonLinearMGBoostEmulator()

# Print a header
print("Parameter Ranges for NonLinearMGBoostEmulator (e-MANTIS):")
print("=" * 50)

# Retrieve the parameter ranges
params_range = emu.params_range

# Display the parameters and their ranges with organized formatting
print(f"{'Parameter':<30} {'Min Value':<15} {'Max Value':<15} {'Type'}")
print("-" * 50)

for param, bounds in params_range.items():
    min_value = bounds['min_value']
    max_value = bounds['max_value']

    # Check if the parameter is continuous or discrete
    if isinstance(min_value, (int, float)) and isinstance(max_value, (int, float)):
        if min_value == max_value:
            param_type = 'Discrete'  # A single value would be considered discrete
        else:
            param_type = 'Continuous'
    else:
        param_type = 'Discrete'

    print(f"{param:<30} {min_value:<15} {max_value:<15} {param_type}")
```
